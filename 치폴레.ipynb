### Step 5. item_price에서 $를 없애고 float 형식으로 새롭게 저장해주세요.

```python
def remove_sign(x):
    return float(x[1:])

chip['item_price']=chip['item_price'].apply(remove_sign)
```

### Step 6. 아이템별 주문 수를 출력해주세요

### Step 7. 아이템별 주문 총량을 출력해주세요.

### Step 8. 각 아이템의 가격을 구해서 unit_price 컬럼에 추가해주세요.
주문당 quantity가 1일때의 가격 (item_price / quantity)

```python
chip['unit_price']=chip['item_price']/chip['quantity']
```

### Step 9. 아이템별 개당 가격의 평균을 구하고 가장 비싼순으로 5개를 출력해주세요

```python
largest5 =sorted(chip.groupby('item_name')['unit_price'].mean(),reverse=True)[:5]
largest5
```

```python
result =chip.groupby('item_name')['unit_price'].mean()
result.sort_values(ascending=False).head()
```
