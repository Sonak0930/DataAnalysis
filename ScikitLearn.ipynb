```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score

import numpy as np
import pandas as pd

iris=load_iris()
features= pd.DataFrame(iris.data)
label= pd.Series(iris.target)

dt_clf=DecisionTreeClassifier(random_state=156)
```

```python
kfold=KFold(n_splits=5,shuffle=True)
#test split에서 이미 랜덤 생성이 되었으므로,섞지는 않는다.
cv_accuracy=[]
n_iter=0

#현재 KFOld 를 5등분 해줘.
#testindex: 0-29까지 30개 검증데이터=fold1 설정.

#5등분 학습 * 5회.
for train_index, test_index in kfold.split(features):
    X_train=features.iloc[train_index]
    X_test=features.iloc[test_index]
    y_train,y_test=label.iloc[train_index],label.iloc[test_index]
    
    dt_clf.fit(X_train,y_train)
    pred=dt_clf.predict(X_test)
    n_iter+=1
    
    accuracy=np.round(accuracy_score(y_test,pred),4)
    train_size=X_train.shape[0]
    test_size=X_test.shape[0]
    
    print('# 검증세션 ',n_iter)
    print('교차 검증 정확도 :',accuracy)    
    print('학습데이터 크기 :',train_size)
    print('검증데이터 크기 :',test_size)
    print('검증 인덱스 :',test_index)
    cv_accuracy.append(accuracy)

print('평균 검증 정확도 :', np.mean(cv_accuracy))
```

교차 검증은 학습을 위한 목표가 아니라,
학습 자체가 유효한지 테스트하기 위함이다.

-> 학습을 할 때마다 정확도가 떨어짐.

```python
from sklearn.model_selection import cross_val_score

iris_data = load_iris()
dt_clf = DecisionTreeClassifier(random_state= 156)
data = iris_data.data
label = iris_data.target

#cv= num of splits
scores = cross_val_score(dt_clf,data,label,scoring='accuracy',cv=5)
print('교차 검증별 정확도:', np.round(scores, 4))
print('평균 검증 정확도:', np.round(np.mean(scores), 4))
```
