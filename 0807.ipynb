```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

#items = [['TV', '냉장고', '전자레인지', '컴퓨터'], ['선풍기', '선풍기', '믹서', '냉장고']]
items=[[1,2,3],[4,5,6]]
print(items)

#2차원 배열로 넣어줘야한다.
#reshape에는 -1을 넣으면 아무거나 들어와도 됨.
#단 parameter의 곱은 전체 shape의 크기와 같아야함.
#따라서 -1,1인 경우, size=6이기 때문에 6*1로 변경됨.
items = np.array(items).reshape(-1,1)
print(items.shape)
print(items)
oh_encoder = OneHotEncoder()
oh_encoder.fit(items)
oh_labels = oh_encoder.transform(items)
```

```python
import pandas as pd
pd.DataFrame(oh_labels.toarray(),columns=oh_encoder.categories_)

#범주형
#명목형 데이터(가전제품 목록)
#bool로 바꿔주는 이유: 값의 크기에 따른 순서가 없기 때문.
#순서형?  (10대,20대, 등급)


#one hot encoding을 사용하는이유가 뭘까?(숙제)
#독립변수(명목형)을 숫자로 변환하기 위함..인데 다른 데이터는 안되는건가?
```
