```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np
import pandas as pd
df=pd.read_csv('data/penguin.csv')
species=df['species']

species=np.array(species).reshape(-1,1)

oh_encoder=OneHotEncoder()
oh_encoder.fit(species)
oh_labels=oh_encoder.transform(species)

oh_labels.toarray()

```

```python
df2=pd.DataFrame({'item':['Adelie','Gentoo','Chinstrap']})

dummies=pd.get_dummies(df2)

df=pd.concat([df,dummies],axis=1)


df
```

reshape(-1,1)을 하는 이유.
1. 2차원 매트릭스로 만들어야 곱하기가 가능하기 때문.
2. Feature와 sample을 구분하기 위함(다른 데이터셋 통용)

```python
import matplotlib.pyplot as plt
import seaborn as sb
from scipy import stats



features=['flipper_length_mm','bill_length_mm','item_Adelie','item_Chinstrap','item_Gentoo']

for feature in features:
    sb.regplot(x=feature,y='body_mass_g',data=df)
    print(df[feature].corr(df['body_mass_g']))
```
