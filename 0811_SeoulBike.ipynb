```python
def check_day(row):

    condition1=row['Week'] in ['Saturday','Sunday']
    condition2=row['Holiday']=='Holiday'
    if condition1 or condition2:
        return 'Yes'
    else:
        return 'No'

bike['Day off']=bike.apply(check_day,axis=1)
bike['Day off'].value_counts()/24
```

```python
# drop Functioning day 중 rent=0

#preprocessing
# 삭제: 이슬점, 날짜, Functioning day 중 rent=0
bike = bike[bike['Rented Bike Count'] >0 ]

X= bike.drop(['Date','Dew point temperature(C)','Functioning Day','Holiday','Week','Rented Bike Count'],axis=1)
y=bike['Rented Bike Count']
```

```python
# encoding
```

```python
# onehot encoding
from sklearn.preprocessing import OneHotEncoder

oh_encoder=OneHotEncoder(sparse_output=False)
mat =oh_encoder.fit_transform(X[['Hour','Seasons','Day off']])


oh_X= pd.DataFrame(mat,columns=oh_encoder.get_feature_names_out())

oh_X
```

```python
#scaling
from sklearn.preprocessing import StandardScaler
st_scaler=StandardScaler()

numeric_columns=['Temperature(C)', 'Humidity(%)', 'Wind speed (m/s)',
       'Visibility (10m)', 'Solar Radiation (MJ/m2)', 'Rainfall(mm)',
       'Snowfall (cm)']
st_res=st_scaler.fit_transform(X[numeric_columns])
st_X=pd.DataFrame(st_res,columns=numeric_columns)


X_scaled=pd.concat([st_X,oh_X],axis=1)

X_scaled
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeRegressor 


dt_reg=DecisionTreeRegressor()
cross_val_score(dt_reg,X_scaled,y,scoring='neg_mean_squared_error',cv=10)
```

```python
from sklearn.model_selection import cross_val_score ,train_test_split
from sklearn.tree import DecisionTreeRegressor 

#학습 전에 데이터를 좀 섞어도(샘플링 해도) 검증 완료.
X_rand=X_scaled.sample(frac=1,random_state=0)
y_rand=y.sample(frac=1,random_state=0)

dt_reg=DecisionTreeRegressor()
cross_val_score(dt_reg,X_rand,y_rand,scoring='neg_mean_squared_error',cv=10)
```

```python
#Model Learning and prediction
# 일단 3개 모델 중 가장 성능이 좋은 친구를 찾아보자

from sklearn.model_selection import train_test_split

X_train, X_test, y_train,y_test=train_test_split(X_scaled,y,test_size=0.2,random_state=0)
print(X_train.shape,X_test.shape,y_train.shape,y_test.shape)

```

```python
#Linear model: 

from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error,r2_score

lr_reg=LinearRegression()
lr_reg.fit(X_train,y_train)
pred=lr_reg.predict(X_test)
print(root_mean_squared_error(y_test,pred))
print(r2_score(y_test,pred))


y_test.describe()

#r2스코어가 뭐였더라??
```

```python
#ridge를 했으나 큰 차이가 없었다.

from sklearn.linear_model import Ridge

ridge=Ridge(alpha=0.1)
ridge.fit(X_train,y_train)
pred=ridge.predict(X_test)
print(root_mean_squared_error(y_test,pred))
print(r2_score(y_test,pred))


y_test.describe()
```

```python
from sklearn.ensemble import RandomForestRegressor


rf_reg=RandomForestRegressor()
rf_reg.fit(X_train,y_train)
pred=rf_reg.predict(X_test)
print(root_mean_squared_error(y_test,pred))
print(r2_score(y_test,pred))

```

```python
from sklearn.model_selection import GridSearchCV

rf_reg=RandomForestRegressor(n_estimators=200)

parameters={
    'max_depth':[None,3,5,7],
    #1 leaf당 10개 이상은 두자.
    'min_samples_split':[2,5,9],
    #1리프당 더이상 자르지 않을 데이터 개수.
    'min_samples_leaf':[1,5,10],
    'random_state':[0]
}
grid=GridSearchCV(rf_reg,param_grid=parameters,scoring='neg_root_mean_squared_error',cv=5)

grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
best_model=grid.best_estimator_
pred= best_model.predict(X_test)

print(root_mean_squared_error(y_test,pred))
print(r2_score(y_test,pred))
```
